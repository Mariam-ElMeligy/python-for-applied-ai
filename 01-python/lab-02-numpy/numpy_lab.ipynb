```python
import numpy as np
```

### Part 1 — Understand the Array

```python
results = np.array([
    [0.82, 0.78, 0.80],
    [0.91, 0.88, 0.90],
    [0.76, 0.81, 0.79],
    [0.88, 0.84, 0.86],
    [0.69, 0.72, 0.70]
])
```

##### Q1

```python
print(results)
```

```python
print("Array Properties:")
print(f"Shape: {results.shape}") 
print(f"No of Dimentions: {results.ndim}")
print(f"Size: {results.size}")
print(f"Data Type: {results.dtype}")
```

Explanations:

* Shape: (5, 3) -> the dataset consists of 5 rows/samples (5 Experiments) & 3 features (Accuracy,
 Precision, Recall)

* No of Dimentions: 2 -> the dataset represents a 2D matrix (X,Y)

* Size: 15 -> The total measurments in the dataset = 15 measurement

* Data Type: float64 -> the dataset measurements are of type float (single-precision floating-point)

##### Q2

```python
# Converting all metric values to percentages:
print(results * 100)
```

### Part 2 — Select & Reshape

##### Q3

```python
# (a) 
print(f"a. E2 Recall: {results[1][2]}")
# (b)
print(f"b. Accuracy: {results[:,0]}")
# (c)
print("c. Accuracy and Precision for the First Three Experiments:")
print(results[0:3,0:2])
```

##### Q4

```python
arr = np.arange(24)

print(f"array: {arr}")
print(f"shape: {arr.shape}") # Vector of 24 Elements
print(f"size: {arr.size}")
```

```python
# Reshaping
arr_reshaped = arr.reshape(4, 6)

print("reshaped array:")
print(arr_reshaped)
print(f"reshaped array shape: {arr_reshaped.shape}")
print(f"reshaped array size: {arr_reshaped.size}")
```

```python
# Flattening back
arr_flattened = arr_reshaped.ravel()
print(f"flattened array: {arr_flattened}")
print(f"flattened array shape: {arr_flattened.shape}")
print(f"flattened array size: {arr_flattened.size}")
```

NOTE: since the 1D vector (24,) and 2D matrix (4, 6) share the same size = 24, so the reshaping is Valid

### Part 3 — Analyze the Results

##### Q5

```python
# (a)
print(f"1. Overall Mean = {results.mean()}")
# (b)
print(f"2. Mean of each metric = {results.mean(axis=0)}")
# (c)
print(f"3. Mean score of each experiment = {results.mean(axis=1)}")
```

Explanation:
1. Overall Mean = 0.81 -> "results.mean()" without specifying a certain axis renders the overall average of all matrix       
( average of all Experiment dataset measurements )

2. Mean of each metric = [0.812 0.806 0.81 ] -> "results.mean(axis=0)" renders the mean of each column in the matrix            
( average of each Metric "Accuracy, Precision, Recall")

3. Mean score of each experiment = [0.8 0.897 0.787 0.86 0.703] -> "results.mean(axis=1)" renders the mean of each row in the matrix            
( average of each Experiment "E1, E2, E3, E4, E5")

### Part 4 — Filter & Extend the Dataset

##### Q7

```python
# Filtering
filtered_results = results[(results[:, 0] >= 0.80) & (results[:, 2] >= 0.80)]

print("Filtered Results:")
print(filtered_results)
print(f"filtered results has {filtered_results.shape[1]} colums")
```

##### Q8

```python
# Vertical Stacking
new_experiment = np.array([[0.85, 0.83, 0.87]])
print(f"New Experiment (E6): {new_experiment}")

results = np.vstack((results, new_experiment))
print("Updated Results with new Experiment (E6):")
print(results)
```

##### Q9

```python
# Horizontal Stacking
f1 = np.zeros((results.shape[0], 1))
print("New Metric F1:")
print(f1)

results = np.hstack((results, f1))
print("Updated Results with new Metric (F1):")
print(results)
```
